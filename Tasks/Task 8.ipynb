```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from tabulate import tabulate

# Load dataset
df = pd.read_csv('customer segmentation.csv')

# Select relevant features for clustering (assuming all except ID or non-numeric columns)
# Adjust column list based on actual dataset structure
features = df.select_dtypes(include=['float64', 'int64']).columns
X = df[features].values

# Number of clusters
k = 3

# Initialize and fit KMeans
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans.fit(X)

# Cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Calculate silhouette score
sil_score = silhouette_score(X, labels)

# Print results
print(f"K-Means Clustering with k = {k}\n")
print("Cluster Labels for each data point:")
print(labels)

print("\nCluster Centroids:")
print(tabulate(centroids, headers=features, tablefmt='grid'))

print(f"\nSilhouette Score: {sil_score:.4f}")
```
